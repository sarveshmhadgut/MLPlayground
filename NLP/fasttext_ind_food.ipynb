```python
import pandas as pd
```

```python
df = pd.read_csv("./datasets/indian_food.csv")
df
```

```python
df = df.drop(
    columns=[
        "TranslatedIngredients",
        "URL",
        "image-url",
        "Ingredient-count",
        "TotalTimeInMins",
    ]
)
df
```

```python
import re
import spacy

nlp = spacy.load("en_core_web_sm")


def preprocess(text):
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"[ \n]+", " ", text)
    doc = nlp(text)

    text = " ".join([str(y.lemma_) for y in doc if not y.is_stop])
    return text.strip().lower()
```

```python
df["preprocessed_instructions"] = df["TranslatedInstructions"].apply(preprocess)
df["preprocessed_instructions"]
```

```python
df["preprocessed_instructions"].to_csv(
    "./datasets/indian_food_ins.text", header=False, index=False
)
```
