```python
# let import necessary libraries and create the object

# for nltk
import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

# downloading all neccessary packages related to nltk
nltk.download("all", quiet=True)


# for spacy
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
# using stemming in nltk
lst_words = [
    "running",
    "painting",
    "walking",
    "dressing",
    "likely",
    "children",
    "whom",
    "good",
    "ate",
    "fishing",
]

stems = [stemmer.stem(word) for word in lst_words]
print(f"\t{'Word':<15}{'Stem':<15}\n")
for word, stem in zip(lst_words, stems):
    print(f"\t{word:<15}{stem:<15}")
```

```python
# using lemmatization in spacy

doc = nlp("running painting walking dressing likely children who good ate fishing")

lemmas = [word.lemma_ for word in doc]

print(f"\t{'Word':<15}{'Lemma':<15}\n")
for word, lemma in zip(lst_words, lemmas):
    print(f"\t{word:<15}{lemma:<15}")
```

```python
text = """
Latha is very multi talented girl.She is good at many skills like dancing, running, singing, playing. She also likes eating Pav Bhagi. she has a habit of fishing and swimming too.Besides all this, she is a wonderful at cooking too.
"""
```

```python
# using stemming in nltk
# step1: Word tokenizing
tokens = [word.strip() for word in text.split()]

# step2: getting the base form for each token using stemmer
stems = [stemmer.stem(token) for token in tokens]

# step3: joining all words in a list into string using 'join()'
" ".join(stems)
```

```python
# using lemmatisation in spacy
# step1: Creating the object for the given text
doc = nlp(text)

# step2: getting the base form for each token using spacy 'lemma_'
lemmas = [str(token.lemma_).strip() for token in doc]

# step3: joining all words in a list into string using 'join()'
" ".join(lemmas).strip()
```
