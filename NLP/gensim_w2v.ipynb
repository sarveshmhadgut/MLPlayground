```python
import gensim
import pandas as pd
```

```python
df = pd.read_json("datasets/reviews_Sports_and_Outdoors_5.json", lines=True)
df
```

```python
from gensim.utils import simple_preprocess

reviews = df["reviewText"].apply(simple_preprocess)
reviews
```

```python
reviews.iloc[0]
```

```python
from gensim.models import Word2Vec

model = Word2Vec(window=5, min_count=5)
```

```python
model.build_vocab(reviews, progress_per=5000)
```

```python
model.train(reviews, total_examples=model.corpus_count, epochs=5)
```

```python
model.wv.most_similar("fabulous")
```

```python
model.wv.similarity(w1="good", w2="phenomenal")
```
