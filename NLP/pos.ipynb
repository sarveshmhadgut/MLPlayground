```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
with open("./datasets/news_story.txt", "r") as f:
    news_text = f.read()

news_text[:500]
```

```python
doc = nlp(news_text)
pos_tags = dict()
for token in doc:
    pos_tags[str(token)] = str(str(token.pos_))
print(pos_tags)
```

```python
nouns = []
for k, v in pos_tags.items():
    if v == "NOUN":
        nouns.append(k)
print(nouns)
```

```python
nums = []
for k, v in pos_tags.items():
    if v == "NUM":
        nums.append(k)
print(nums)
```

```python
pos_freq = doc.count_by(spacy.attrs.POS)
pos_freq
for k, v in pos_freq.items():
    print(f"{doc.vocab[int(k)].text:<15}, {v:<15}")
```
