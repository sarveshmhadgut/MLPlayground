```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("HandleMissing").getOrCreate()
sc = spark.sparkContext
sc.setLogLevel("ERROR")
spark
```

```python
df = spark.read.csv(
    "datasets/students_info.csv", header=True, inferSchema=True, nullValue="NULL"
)
df.toPandas()
```

```python
df.printSchema()
```

```python
df.describe().toPandas()
```

### Default Drop

```python
df.na.drop().toPandas()
```

### Drop (any)

```python
drop_any = df.na.drop(how="any")
print(len(df.toPandas()), " -> ", len(drop_any.toPandas()))
drop_any.toPandas()
```

### Drop (all)

```python
drop_all = df.na.drop(how="all")
print(len(df.toPandas()), " -> ", len(drop_all.toPandas()))
drop_all.toPandas()
```

### Drop (w/ threshold)

```python
drop_threshold = df.na.drop(how="any", thresh=8)
print(len(df.toPandas()), " -> ", len(drop_threshold.toPandas()))
drop_threshold.toPandas()
```

### Drop (subset)

```python
df_subset = df.na.drop(how="any", subset=["gpa"])
print(len(df.toPandas()), " -> ", len(df_subset.toPandas()))
df_subset.toPandas()
```

```python
df.na.fill(0, ["gpa"]).toPandas()
```

```python
from pyspark.ml.feature import Imputer

cols = ["age", "attendance_pct"]

imputer = Imputer(
    inputCols=cols, outputCols=[f"temp_{col}" for col in cols], strategy="median"
)
```

```python
df_imputed = imputer.fit(df).transform(df)
df_imputed.select(cols + [f"temp_{col}" for col in cols]).toPandas()
```
