```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("FilterOps").getOrCreate()
sc = spark.sparkContext
sc.setLogLevel("ERROR")
spark
```

```python
df = spark.read.csv(
    "datasets/students_info.csv", header=True, inferSchema=True, nullValue="NULL"
)
df.toPandas()
```

```python
df.filter(df["age"] == 21).toPandas()
```

# Filter

```python
df.filter(df["department"] != "CIVIL").toPandas()
```

```python
df.filter((df["gpa"] > 8) & (df["attendance_pct"] > 70)).select(
    ["name", "gpa", "attendance_pct"]
).toPandas()
```

```python
df.filter(~((df["department"] == "CSE") & (df["gender"] == "M"))).select(
    ["name", "department", "gender"]
).toPandas()
```

# Group and Aggregate

```python
df.groupBy("department").count().toPandas()
```

```python
df.select(["department", "gpa"]).groupBy("department").avg().toPandas()
```

```python
df.groupBy("year").count().toPandas()
```

```python
df.agg({"age": "avg"}).toPandas()
```

## Ordering & Sorting

```python
df.orderBy(df["gpa"].desc()).select(["name", "gpa"]).toPandas()
```

```python
df.sort(["attendance_pct"], ascending=False).select(
    ["name", "attendance_pct"]
).toPandas()
```
